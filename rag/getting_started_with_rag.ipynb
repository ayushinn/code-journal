    <h1 style="color: #fadad9 " > RAG , Retrieval-Augmented Generation ˚.⋆꒰১ ໒꒱⋆.˚ </h1>
     <p style="color: #e9abae " >Rag is an AI technique that improves large language model outputs by checking external knowledge bases before answering. </p>

</div>



<div style="background-color: #1e1e1e; padding: 16px; border-radius: 8px; border: 1px solid #333333; font-family: monospace;">
    <h3 style="color: #fadad9">How RAG works? 𑣲𝄞</h3>
     <p style="color: #e9abae " > 
     ⚝ Retrieval: The system searches an external database or document collection for text related to a user's prompt.<br>
     ⚝ Augmentation: It combines the retrieved facts with the original user query into a single context-rich prompt.<br>
     ⚝ Generation: The language model reads the combined prompt and creates a grounded, factual response.</p>

</div>

<div style="background-color: #1e1e1e; padding: 16px; border-radius: 8px; border: 1px solid #333333; font-family: monospace;">
    <h3 style="color: #fadad9">Benefits of RAG 𑣲𝄞</h3>
     <p style="color: #e9abae " > 
     ⚝ Reduces Hallucinations: Answers rely on verified source documents instead of model guesses.<br>
     ⚝ Fresh Data Access: Connects models to real-time or frequently updated information without expensive retraining.<br>
     ⚝ Domain Specificity: Tailors general AI models to private company data, medical records, or legal documents.</p>
